# Movie Rating Prediction - Exploratory Data Analysis

This notebook explores the Movie dataset and analyzes factors influencing ratings.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

df = pd.read_csv('../data/movies.csv')
print(f"Dataset shape: {df.shape}")
df.head()
```

## Dataset Overview

```python
print("Dataset Info:")
print(df.info())
print("\nBasic Statistics:")
print(df.describe())
```

```python
# Rating distribution
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Get rating column (could be named 'rating', 'imdb_rating', etc.)
rating_col = [col for col in df.columns if 'rating' in col.lower()][0] if any('rating' in col.lower() for col in df.columns) else df.columns[-1]

axes[0].hist(df[rating_col], bins=30, color='#ff7f0e', edgecolor='black')
axes[0].set_xlabel('Rating')
axes[0].set_ylabel('Count')
axes[0].set_title('Rating Distribution')

axes[1].boxplot(df[rating_col])
axes[1].set_ylabel('Rating')
axes[1].set_title('Rating Box Plot')

plt.tight_layout()
plt.show()

print(f"Mean Rating: {df[rating_col].mean():.2f}")
print(f"Median Rating: {df[rating_col].median():.2f}")
print(f"Rating Range: {df[rating_col].min():.1f} - {df[rating_col].max():.1f}")
```

## Feature Analysis

```python
# Budget vs Rating
if 'budget' in df.columns:
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))
    
    axes[0, 0].scatter(df['budget'], df[rating_col], alpha=0.5, color='#ff7f0e')
    axes[0, 0].set_xlabel('Budget')
    axes[0, 0].set_ylabel('Rating')
    axes[0, 0].set_title('Budget vs Rating')
    
    # Runtime vs Rating
    if 'runtime' in df.columns:
        axes[0, 1].scatter(df['runtime'], df[rating_col], alpha=0.5, color='#1f77b4')
        axes[0, 1].set_xlabel('Runtime (minutes)')
        axes[0, 1].set_ylabel('Rating')
        axes[0, 1].set_title('Runtime vs Rating')
    
    # Year vs Rating
    if 'release_year' in df.columns:
        year_rating = df.groupby('release_year')[rating_col].mean()
        axes[1, 0].plot(year_rating.index, year_rating.values, marker='o', color='#2ca02c')
        axes[1, 0].set_xlabel('Release Year')
        axes[1, 0].set_ylabel('Average Rating')
        axes[1, 0].set_title('Rating Trend Over Years')
    
    axes[1, 1].axis('off')
    
    plt.tight_layout()
    plt.show()
```

## Correlation Analysis

```python
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
correlation = df[numeric_cols].corr()[rating_col].sort_values(ascending=False)

plt.figure(figsize=(10, 6))
correlation.plot(kind='barh', color='#1f77b4')
plt.xlabel('Correlation with Rating')
plt.title('Feature Correlation with Movie Rating')
plt.axvline(x=0, color='black', linestyle='-', linewidth=0.5)
plt.tight_layout()
plt.show()

print(correlation)
```

## Key Insights

1. Rating distribution shows the typical patterns of movie ratings
2. Budget and ratings have a moderate positive correlation
3. Runtime shows an interesting relationship with ratings
4. High-voter movies tend to be more stable in rating
5. Genre significantly impacts rating patterns
